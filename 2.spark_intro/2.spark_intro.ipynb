```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName('WordCountDemo') \
    .master('yarn') \
    .getOrCreate()
```

```python
data = [
    "Goku Vegeta Gohan",
    "Goku Frieza Goku",
    "Vegeta Goku Frieza Gohan",
    "Gohan Frieza Goku Goku"
]
```

```python
rdd = spark.sparkContext.parallelize(data)
```

```python
rdd.collect()
```

```python
hdfs_path = '/tmp/inputhdfsdbz.txt'

rdd_hdfs = spark.sparkContext.textFile(hdfs_path)
```

```python
rdd_hdfs.collect()
```

```python
!hadoop fs -ls /tmp/
```

```python
rdd_hdfs_mapped = rdd_hdfs.map(lambda line:line.split(" "))
```

```python
rdd_hdfs_mapped.collect()
```

#### filter operation

```python
filter_rdd_hdfs = rdd_hdfs_mapped.filter(lambda words: "Goku" in words)
```

```python
filter_rdd_hdfs.collect()
```

```python
filter_rdd_hdfs = rdd_hdfs_mapped.filter(lambda words: "Frieza" in words)
```

```python
filter_rdd_hdfs.collect()
```

```python
filter_rdd_hdfs = rdd_hdfs_mapped.filter(lambda words: "Vegeta" in words)
```

```python
filter_rdd_hdfs.collect()
```

```python
spark.stop()
```

### Why is Spark Lazy?

```python
hdfs_path = '/tmp/customers_500mb.csv'
rdd_big_file = spark.sparkContext.textFile(hdfs_path)
```

```python
big_ans_combined = rdd_big_file.flatMap(lambda line: line.split(' ')).map(lambda word:(word,1)).reduceByKey(lambda a,b : a+b)
```
